```python
import yfinance as yf
ticker = "VWS.CO"

data = yf.Ticker(ticker)
data.get_fast_info()
table = data.history(period="5d").drop(columns=["Dividends", "Stock Splits"])

# extract the last 5 days of closing prices
closing_prices = data.history(period="5d")["Close"].tolist()
opening_prices = data.history(period="5d")["Open"].tolist()
low_prices = data.history(period="5d")["Low"].tolist()
high_prices = data.history(period="5d")["High"].tolist()

# extract dates as strings in the format "YYYY-MM-DD"
dates = table.index.strftime("%Y/%m/%d").tolist()
table.index[:], dates, closing_prices, opening_prices, low_prices, high_prices

price_observations = {}

for date in dates:
    price_observations[date] = {
        "closing_price": closing_prices.pop(0),
        "opening_price": opening_prices.pop(0),
        "low_price": low_prices.pop(0),
        "high_price": high_prices.pop(0),
    }
price_observations
```
